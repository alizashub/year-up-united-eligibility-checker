## 🎯 Is **"YEAR UP UNITED"** Right for You?
Created by Aliza Rashid 👩‍💻

> This small interactive program asks a few questions to see if Year Up United might be a good fit for you. Just click the ▶️ Run button on the right or the Run all button above, then answer the questions to find out!

```python
print("Welcome! Let's see if you are eligible for Year Up United.\n")

eligible = True

# Age Check
while True:
    age_input = input("Please enter your age as a whole number (e.g., 21):").strip()
    if not age_input.isdigit():
         print("⚠️ Oops! That doesn’t look like a number. Please enter your age using digits only (e.g., 21).")
         continue

    age = int(age_input)
    if 18 <= age <= 29:
        break
    else:
        print("\nSorry, you are not eligible for Year Up United.")
        print("You must be between 18 and 29 years old for eligibility.")
        eligible = False
        break

# Work Authorization Check

if eligible:
    while True:
        work_auth = input("Are you authorized to work in the U.S. for the full program duration? (Yes/No): ").strip().upper()
        if work_auth in ["YES", "Y"]:
            break
        elif work_auth in ["NO", "N"]:
            daca = input("Do you have DACA status? (Yes/No): ").strip().upper()
            if daca in ["YES", "Y"]:
                break
            elif daca in ["NO", "N"]:
                print("\nSorry, you must be authorized to work in the U.S. or have DACA status to be eligible.")
                eligible = False
                break
            else:
                print("⚠️ Please answer Yes or No.")
        else:
            print("⚠️ Please answer Yes or No.")


# Secondary Education Check

if eligible:
    while True:
        secondary = input("Do you have a high school diploma or GED? (Yes/No): ").strip().upper()
        if secondary in ["YES", "Y"]:
            break
        elif secondary in ["NO", "N"]:
            print("\nSorry, a secondary education (High School or GED) is required.")
            eligible = False
            break
        else:
            print("⚠️ Please answer Yes or No.")

# College Degree Check

if eligible:
    while True:
        college = input("Have you obtained a Bachelor's degree? (Yes/No): ").strip().upper()
        if college in ["NO", "N"]:
            break
        elif college in ["YES", "Y"]:
            print("\nSorry, this program is for applicants without a college degree.")
            eligible = False
            break
        else:
            print("⚠️ Please answer Yes or No.")


# Income Check

if eligible:
    while True:
        income = input("Is your income considered low to moderate? (Yes/No): ").strip().upper()
        if income in ["YES", "Y"]:
            break
        elif income in ["NO", "N"]:
            print("\nSorry, you must have a low to moderate income for eligibility.")
            eligible = False
            break
        else:
            print("⚠️ Please answer Yes or No.")


# Availability Check

if eligible:
    while True:
        available = input("Are you available full-time (5 days a week, Monday to Friday)? (Yes/No): ").strip().upper()
        if available in ["YES", "Y"]:
            break
        elif available in ["NO", "N"]:
            print("\nSorry, full-time availability is required.")
            eligible = False
            break
        else:
            print("⚠️ Please answer Yes or No.")

# Location Check

if eligible:
    program_cities = ["ARIZONA", "AUSTIN", "BALTIMORE", "BAY AREA", "CHARLOTTE",
                      "CHICAGO", "DALLAS", "FORT WORTH", "ATLANTA", "BOSTON",
                      "PHILADELPHIA", "JACKSONVILLE", "LOS ANGELES", "MICHIGAN",
                      "WASHINGTON D.C.", "NEW JERSEY", "NEW YORK", "PITTSBURGH",
                      "PUGET SOUND", "RHODE ISLAND", "SOUTH FLORIDA", "TAMPA BAY",
                      "WILMINGTON"]

    print("\nAvailable locations:")
    for location in sorted(program_cities):
        print("- " + location)

    while True:
        commute = input("\nAre you within commuting distance of one of these locations? (Yes/No): ").strip().upper()
        if commute in ["YES", "Y"]:
            print("Great! You are within commuting distance. ✅")
            break
        elif commute in ["NO", "N"]:
            print("\nSorry, you must live within commuting distance of a program location.")
            eligible = False
            break
        else:
            print("⚠️ Please answer Yes or No.")

# Final Eligibility Message

print("\n------------------------")
if eligible:
    print("🎉 Congratulations! You meet all the eligibility criteria!")
    print("Exciting opportunities await you. Check upcoming cohort dates here:")
    print("https://www.yearup.org/")
else:
  print("You don’t meet the eligibility criteria for Year Up United. 🩷 Keep exploring — the right opportunities are out there for you!")

print("------------------------")
```
